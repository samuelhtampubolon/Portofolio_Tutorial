Unsupervised Learning (Clustering and PCA (Principal Component Analysis))

```python
# Loading the Unsupervised Data
# In scikit-learn, make_blobs is a function used to generate synthetic 'blobs'
# synthetic 'blobs' of data—clusters of points that follow a Gaussian (normal) distribution
from sklearn.datasets import make_blobs
# Generating the Clusters
X_blob, y_blob = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)
# Plotting Raw Clusters
plt.scatter(X_blob[:, 0], X_blob[:, 1])
plt.show()
```

```python
# Importing the K-Means
from sklearn.cluster import KMeans
# Instantiating the K-Means
kmeans = KMeans(n_clusters=4)
# Training the K-Means
kmeans.fit(X_blob)
# Extracting the Cluster Centers
centers = kmeans.cluster_centers_
# Extracting Labels
labels = kmeans.labels_
# Plotting the Clustered Data
plt.scatter(X_blob[:, 0], X_blob[:, 1], c=labels, cmap='viridis')
```

```python
# Plotting the Cluster Centers
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.5)
plt.show()
```

```python
# Calculating the Inertia
# Writing a loop testing K from 1 to 10 and saving kmeans.inertia_ for each
kmeans.inertia_
```

```python
# Plotting the Elbow Graph
# Plotting the inertia values against K to visually find the optimal number of clusters
# Importing the PCA
from sklearn.decomposition import PCA
# Instantiating the PCA
pca = PCA(n_components=2)
# Transforming the Data
# to reduce the breast cancer dataset to 2 dimensions
X_pca = pca.fit_transform(X_cls)
print(X_pca.shape)
print(X_cls.shape)
# viridis is a color palette in Python data visualization library
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_cls, cmap='viridis')
plt.show()
```

```python
# Instantiating another PCA
pca3 = PCA(n_components=3)
# Transforming the Data to reduce the breast cancer dataset to 3 dimensions
X_pca3 = pca3.fit_transform(X_cls)
print(X_pca3.shape)
print(X_cls.shape)

# Creating 3 classifications/clusters using KMeans
from sklearn.cluster import KMeans
kmeans3 = KMeans(n_clusters=3, random_state=42)
cluster_labels = kmeans3.fit_predict(X_pca3)

# Plotting in 3 Dimensions
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# turbo is a color palette in Python data visualization library
# Plotting X_pca3 using 3 components and coloring by our 3 cluster labels
scatter = ax.scatter(X_pca3[:, 0], X_pca3[:, 1], X_pca3[:, 2], c=cluster_labels, cmap='turbo', s=40, alpha=0.8)

ax.set_title('3D PCA with 3 Clusters')
ax.set_xlabel('Principal Component 1')
ax.set_ylabel('Principal Component 2')
ax.set_zlabel('Principal Component 3')
plt.colorbar(scatter, label='Cluster')
plt.show()
```

```python
# Train the ANN on the GPU
# The model allocates memory automatically on the GPU instance
# Epoch refers to one complete pass through the entire training dataset by the
# neural network
history_ann = model_ann.fit(X_train, y_train, epochs=50, validation_split=0.2, callbacks=[early_stop])
```

```python
# The MNIST (Modified National Institute of Standards and Technology) dataset is
# a massive collection of 70,000 grayscale images of handwritten digits Loading
# the MNIST dataset for Convolutional Neural Network (CNN) training This dataset
# contains 70,000 grayscale images of handwritten digits
mnist = tf.keras.datasets.mnist
(X_train_img, y_train_img), (X_test_img, y_test_img) = mnist.load_data()
```

```python
# Normalizing the pixel values to range between 0 and 1
X_train_img = X_train_img / 255.0
X_test_img = X_test_img / 255.0
plt.imshow(X_train_img[0])
```

```python
# A Convolutional Neural Network (CNN) is a class of deep learning models
# specialized in processing grid-structured data like images and video
# Initializing the CNN model
model_cnn = tf.keras.Sequential()
```

```python
# Adding a Convolutional layer to extract spatial features from the images
model_cnn.add(tf.keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)))
```

```python
# Adding a MaxPooling layer to downsample the spatial dimensions
# This reduces computational load and extracts dominant features
model_cnn.add(tf.keras.layers.MaxPooling2D((2, 2)))
```

```python
# Flattening the 2D feature maps into a 1D vector
model_cnn.add(tf.keras.layers.Flatten())
```
